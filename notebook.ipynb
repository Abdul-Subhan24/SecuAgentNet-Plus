## Project Overview
<a id="project-overview"></a>
SecuAgentNet+ demonstrates an enterprise-friendly approach to triaging digital incidents (email, sms, uploads, documents). The system is intentionally conservative (high recall), emits full per-incident audit logs, and enforces governance decisions via an autonomy score and HITL rules.

**Key features**
- Hybrid detection (rule engine + LLM agent)
- Structured JSON outputs and audit trails per incident
- Governance engine (autonomy scoring and HITL)
- Modular agents: Detection, Misuse classification, Deepfake screening, Cross-check, Reporting
