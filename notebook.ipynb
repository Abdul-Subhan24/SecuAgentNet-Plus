1. Introduction

With rising phishing attacks, deepfakes, digital fraud, and content-based cyber risks, organizations need fast and reliable incident analysis.
SecuAgentNet+ is a hybrid AI system that combines:

Traditional Machine Learning (baseline detection model)

LLM-powered reasoning agent (OpenAI / Ollama)

Governance scoring engine (risk prioritization)

Human-in-the-loop compatible framework

The system analyzes incidents (emails, documents, messages, uploads) and classifies them as:

benign

phishing

job scams

invoice fraud

harassment

malicious or suspicious content

It also extracts Indicators of Compromise (IOCs) like URLs and domains.

3. Architecture Overview

The solution uses a three-layer hybrid design:

A. ML Baseline Detector

A lightweight classifier produces an initial threat score (0–1).

B. LLM Detection Agent (Reasoning Layer)

A custom prompt instructs the model to:

identify threat category

extract IOCs

generate a short explanation

output structured JSON

This significantly improves accuracy compared to ML-only detection.

C. Governance & Risk Engine

This layer assigns final risk labels:

High Risk → immediate action

Medium Risk → review

Low Risk → log only

Governance rules ensure consistency and make the system suitable for enterprise environments.

4. Dataset
